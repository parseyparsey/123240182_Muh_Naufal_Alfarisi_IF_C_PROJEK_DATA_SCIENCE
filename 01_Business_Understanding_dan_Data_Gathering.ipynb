## 1. Business Understanding

### 1.1 Latar belakang masalah

**Pernyataan masalah:**  
Bagaimana membangun model klasifikasi multikelas yang memprediksi jenis tutupan hutan berdasarkan variabel topografi dan lingkungan pada dataset Covertype?

1. Memahami karakteristik dan kualitas dataset Covertype.

### 1.5 Metode dan metrik keberhasilan

Metode yang digunakan adalah **klasifikasi multikelas**. Notebook modelling membandingkan baseline sederhana, Decision Tree, dan Random Forest.

Metrik utama adalah **Macro-F1**, karena setiap kelas memperoleh bobot yang sama dalam perhitungan sehingga kelas minoritas tidak tertutupi oleh kelas mayoritas. Metrik pendukung:

- **Accuracy:** proporsi prediksi yang benar secara keseluruhan.
- **Balanced accuracy:** rata-rata recall antar kelas.
- **Precision dan recall per kelas:** untuk melihat kesalahan pada setiap jenis tutupan.
- **Confusion matrix:** untuk melihat pasangan kelas yang sering tertukar.

Tidak ditetapkan ambang performa yang dianggap pasti tercapai sebelum eksperimen dilakukan. Model dibandingkan dengan baseline dan dipilih berdasarkan hasil pengujian, bukan hanya nilai accuracy.

- Satu observasi merepresentasikan petak berukuran 30 × 30 meter menurut dokumentasi dataset.

### 2.2 Gambaran isi dataset

Dataset berisi 55 kolom:

- 10 fitur numerik/topografi.
- 4 fitur indikator `Wilderness_Area` (one-hot).
- 40 fitur indikator `Soil_Type` (one-hot).
- 1 target `Cover_Type`.

Setiap baris adalah observasi petak lahan. File mentah tidak memiliki nama kolom, sehingga nama kolom diberikan sesuai urutan atribut pada dokumentasi Covertype.

**Target `Cover_Type`:**

| Kode | Nama kelas |
|---:|---|
| 1 | Spruce/Fir |
| 2 | Lodgepole Pine |
| 3 | Ponderosa Pine |
| 4 | Cottonwood/Willow |
| 5 | Aspen |
| 6 | Douglas-fir |
| 7 | Krummholz |

### 2.3 Kamus fitur

| Fitur | Makna |
|---|---|
| `Elevation` | Ketinggian dari permukaan laut (meter) |
| `Aspect` | Arah hadap lereng (derajat, 0–360) |
| `Slope` | Kemiringan lereng (derajat) |
| `Horizontal_Distance_To_Hydrology` | Jarak horizontal ke sumber air terdekat (meter) |
| `Vertical_Distance_To_Hydrology` | Jarak vertikal ke sumber air terdekat (meter; dapat negatif) |
| `Horizontal_Distance_To_Roadways` | Jarak horizontal ke jalan terdekat (meter) |
| `Hillshade_9am` | Indeks hillshade pukul 09.00 |
| `Hillshade_Noon` | Indeks hillshade tengah hari |
| `Hillshade_3pm` | Indeks hillshade pukul 15.00 |
| `Horizontal_Distance_To_Fire_Points` | Jarak horizontal ke titik api terdekat (meter) |
| `Wilderness_Area1`–`Wilderness_Area4` | Indikator keanggotaan area wilderness |
| `Soil_Type1`–`Soil_Type40` | Indikator jenis tanah |
| `Cover_Type` | Label target jenis tutupan hutan (1–7) |

### 2.4 Memuat dataset

Simpan file `covtype.data` di folder yang sama dengan notebook atau di subfolder `data/`. Kode berikut membaca file mentah tanpa header dan memberikan nama kolom.

```python
from pathlib import Path
import pandas as pd
import numpy as np

NUMERIC_FEATURES = [
    "Elevation", "Aspect", "Slope", "Horizontal_Distance_To_Hydrology",
    "Vertical_Distance_To_Hydrology", "Horizontal_Distance_To_Roadways",
    "Hillshade_9am", "Hillshade_Noon", "Hillshade_3pm",
    "Horizontal_Distance_To_Fire_Points"
]
WILDERNESS_COLS = [f"Wilderness_Area{i}" for i in range(1, 5)]
SOIL_COLS = [f"Soil_Type{i}" for i in range(1, 41)]
TARGET_COL = "Cover_Type"
COLUMNS = NUMERIC_FEATURES + WILDERNESS_COLS + SOIL_COLS + [TARGET_COL]

DATA_CANDIDATES = [
    Path("covtype.data"),
    Path("data") / "covtype.data",
    Path("../covtype.data"),
    Path("../data") / "covtype.data",
]
DATA_PATH = next((path for path in DATA_CANDIDATES if path.exists()), None)

if DATA_PATH is None:
    raise FileNotFoundError(
        "File covtype.data tidak ditemukan. Letakkan file tersebut di folder "
        "yang sama dengan notebook atau di subfolder 'data/'."
    )

df = pd.read_csv(DATA_PATH, header=None, names=COLUMNS)
print(f"Dataset berhasil dimuat dari: {DATA_PATH.resolve()}")
print(f"Ukuran dataset: {df.shape[0]:,} baris x {df.shape[1]} kolom")
```

### 2.5 Pemeriksaan awal dataset

Jalankan pemeriksaan berikut untuk memastikan struktur data sesuai dengan dokumentasi. Hasil aktual akan berasal dari file `covtype.data` yang digunakan saat notebook dijalankan.

```python
display(df.head())
print("\nUkuran dataset:", df.shape)
print("\nInformasi tipe data:")
df.info()
```

### Interpretasi tampilan data awal

Lima baris pertama memperlihatkan bahwa file berhasil dibaca dengan urutan kolom yang sesuai: fitur topografi, indikator Wilderness Area, indikator Soil Type, lalu `Cover_Type`. Contoh ini hanya berfungsi memeriksa struktur dan format; lima baris pertama tidak dapat dianggap mewakili distribusi seluruh 581.012 observasi.

Nilai `Cover_Type` pada contoh awal sudah berupa kode kelas 1–7. Kode tersebut adalah label kategori, bukan besaran numerik yang memiliki urutan matematis. Karena itu, rata-rata `Cover_Type` pada `describe()` tidak digunakan sebagai interpretasi substantif.

```python
# Statistik awal dan pemeriksaan kelengkapan
display(df.describe().T)

quality_summary = pd.DataFrame({
    "missing_values": df.isna().sum(),
    "missing_percent": (df.isna().mean() * 100).round(4),
    "unique_values": df.nunique()
})
display(quality_summary)

print("Jumlah baris duplikat:", df.duplicated().sum())
print("Jumlah label target:", df[TARGET_COL].nunique())
print("Label target yang ditemukan:", sorted(df[TARGET_COL].dropna().unique()))
```

### Interpretasi statistik dan pemeriksaan kualitas

Output menunjukkan 581.012 nilai non-null untuk setiap kolom. Dengan demikian, seluruh 55 kolom terisi pada file yang diuji. Fitur numerik memiliki satuan dan rentang yang berbeda: `Elevation` diukur dalam meter, `Aspect` dan `Slope` dalam derajat, jarak dalam meter, sedangkan hillshade berupa indeks.

Pemeriksaan `describe()` digunakan untuk memahami skala dan sebaran awal fitur. Nilai rata-rata target `Cover_Type` tidak ditafsirkan sebagai besaran karena target adalah kode kategori nominal.

```python
# Pemeriksaan kelompok fitur one-hot
wilderness_active = df[WILDERNESS_COLS].sum(axis=1)
soil_active = df[SOIL_COLS].sum(axis=1)

print("Jumlah indikator Wilderness_Area aktif per baris:")
display(wilderness_active.value_counts().sort_index())

print("Jumlah indikator Soil_Type aktif per baris:")
display(soil_active.value_counts().sort_index())

print("Nilai unik pada fitur Wilderness_Area:",
      sorted(pd.unique(df[WILDERNESS_COLS].to_numpy().ravel())))
print("Nilai unik pada fitur Soil_Type:",
      sorted(pd.unique(df[SOIL_COLS].to_numpy().ravel())))
```

### Interpretasi validasi fitur one-hot

Output menunjukkan setiap baris memiliki tepat satu indikator aktif pada kelompok `Wilderness_Area` dan tepat satu indikator aktif pada kelompok `Soil_Type`. Nilai yang ditemukan pada kedua kelompok hanya 0 dan 1. Ini mendukung konsistensi struktur one-hot dataset. Pemeriksaan ini penting karena lebih dari satu indikator aktif atau tidak ada indikator aktif dapat menunjukkan ketidaksesuaian struktur data.
